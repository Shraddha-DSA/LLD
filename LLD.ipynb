## Memento Pattern

**PS: Build a Text Editor that supports write(text) and undo**

```python
#Memento
class TextEditorMemento:
    def __init__(self,text):
        self._text=text
    def get_text(self):
        return self._text
```

```python
#Originator
class TextEditor:
    def __init__(self):
        self._text=""
    def write(self,text):
        self._text+=text
    def save(self):
        return TextEditorMemento(self._text)
    def restore(self,memento):
        self._text=memento.get_text()
    def show(self):
        print(self._text)
```

```python
#careTaker
class History:
    def __init__(self):
        self._history=[]
    def push(self,momento):
        self._history.append(momento)
    def pop(self):
        if self._history:
            return self._history.pop()
        return None
```

```python
#Client
editor=TextEditor()
history=History()
editor.write("Hello")
history.push(editor.save())
editor.show()
history.pop()
previous=history.pop()
if previous:
    editor.restore(previous)
editor.show()
```

## Observer Pattern

**Build a Youtube Channel Notification System(When a channel uploads a new video, sll subscriber should aautomatically recieve a notification**

```python
from abc import ABC, abstractmethod
```

```python
#Observer:An interface that every subscriber follows.
class Subscriber(ABC):
    @abstractmethod
    def update(self,video_title):
        pass
```

```python
#Concrete Observer:Actual subscribers
class User(Subscriber):
    def __init__(self,name):
        self.name=name
    def update(self,video_title):
        print(f"{self.name} received notification: {video_title}")
```

```python
#Subject(Observable):The object being observed
class YouTubeChannel:
    def __init__(self):
        self.subscriber=[]
    def subscribe(self,subscriber):
        self.subscriber.append(subscriber)
    def unsubscribe(self,subscriber):
        self.subscriber.remove(subscriber)
    def notify(self,video_title):
        for subscriber in self.subscriber:
            subscriber.update(video_title)
    def upload_video(self,title):
        print(f"\nUploading: {title}")
        self.notify(title)
```

```python
#Client: Creates the channel and subscribers
channel=YouTubeChannel()
alice=User("Alice")
bob=User("Bob")
charlie=User("Charlie")
channel.subscribe(alice)
channel.subscribe(bob)
channel.subscribe(charlie)
channel.upload_video("Observer Pattern Explained!")
```

## Iterator Pattern

```python
from abc import ABC, abstractmethod
#Iterator: Defines how to traverse
class Iterator(ABC):
    @abstractmethod
    def has_next(self):
        pass
    @abstractmethod
    def next(self):
        pass
```

```python
#Concrete Iterator: Implements the traversal logic
class PlaylistIterator(Iterator):
    def __init__(self,songs):
        self.songs=songs
        self.index=0
    def has_next(self):
        return self.index<len(self.songs)
    def next(self):
        if self.has_next():
            song=self.songs[self.index]
            self.index+=1
            return song
        return None
```

```python
#Aggregate(Collections): The collection that creates an iterator
class Playlist:
    def __init__(self):
        self.songs=[]
    def add_song(self,song):
        self.songs.append(song)
    def create_iterator(self):
        return PlaylistIterator(self.songs)
```

```python
#Client: uses the iterator to access elements one by one
playlist=Playlist()
playlist.add_song("Believer")
playlist.add_song("Shape of You")
playlist.add_song("perfect")
iterator=playlist.create_iterator()
while iterator.has_next():
    print(iterator.next())
```

## State Pattern

**PS: An order goes through these states: Placed,Shipped and delivered**

```python
from abc import ABC, abstractmethod
```

```python
#State: Defines the behavior for each state
class OrderState(ABC):
    @abstractmethod
    def next(self,order):
        pass
```

```python
#Concrete States: Each defines what happens next
class PlacedState(OrderState):
    def next(self,order):
        print("Order shipped")
        order.set_state(ShippedState())
class ShippedState(OrderState):
    def next(self,order):
        print("Order delivered")
        order.set_state(DeliveredState())
class DeliveredState(OrderState):
    def next(self,order):
        print("Order is already delivered")
```

```python
#Context: The object whose behavior changes
class Order:
    def __init__(self):
        self.state=PlacedState()
    def set_state(self,state):
        self.state=state
    def next_state(self):
        self.state.next(self)
```

```python
#Client
order=Order()
order.next_state()
order.next_state()
order.next_state()
```

## Mediator Pattern

**PS: Build a chat room: users should send messages through the chat room not individually**

```python
from abc import ABC, abstractmethod
```

```python
#Mediator: defines communication methods.
class ChatMediator(ABC):
    @abstractmethod
    def send(self,message,sender):
        pass
```

```python
#Concrete Mediator: Responsible for forwarding messages
class ChatRoom(ChatMediator):
    def __init__(self):
        self.users=[]
    def add_user(self,user):
        self.users.append(user)
    def send(self,message,sender):
        for user in self.users:
            if user != sender:
                user.receive(message, sender.name)
```

```python
#Colleague: Object that communicate via the mediator
class User:
    def __init__(self,name,mediator):
        self.name=name
        self.mediator=mediator
    def send(self,message):
        print(f"{self.name} sends: {message}")
        self.mediator.send(message,self)
    def receive(self,message,sender):
        print(f"{self.name} received from {sender}: {message}")
```

```python
#Client: creates the chat room and users
chat_room=ChatRoom()
alice=User("Alice",chat_room)
bob=User("Bob",chat_room)
charlie=User("Charlie",chat_room)

chat_room.add_user(alice)
chat_room.add_user(bob)
chat_room.add_user(charlie)
alice.send("Hello Everyone")
```

## Composite Pattern

**PS: Build a File system with file and Folder**

```python
from abc import ABC, abstractmethod
```

```python
#Component: Common Interface
class FileSystemComponent(ABC):
    @abstractmethod
    def show(self):
        pass
```

```python
#Leaf: individual Object
class File(FileSystemComponent):
    def __init__(self,name):
        self.name=name
    def show(self):
        print(f"File: {self.name}")
```

```python
#Composite: Group Object
class Folder(FileSystemComponent):
    def __init__(self,name):
        self.name=name
        self.children=[]
    def add(self,component):
        self.children.append(component)
    def remove(self,component):
        self.children.remove(component)
    def show(self):
        print(f"Folder: {self.name}")
        for child in self.children:
            child.show()
```

```python
#Client: Works with File and Folder through the same interface
root=Folder("Root")
file1=File("resume.pdf")
file2=File("photo.jpg")
docs=Folder("Document")
docs.add(file1)
images=Folder("Images")
images.add(file2)
root.add(docs)
root.add(images)
root.show()
```

## Chain of Responsibility

```python
#Handler
class SupportHandler(ABC):
    def __init__(self):
        self.next_handler=None
    def set_next(self,handler):
        self.next_handler=handler
        return handler
    @abstractmethod
    def handle(self,request):
        pass
```

```python
#Concrete Handlers
class Level1Support(SupportHandler):
    def handle(self,request):
        if request=="password":
            print("Level 1 handled Password Reset")
        elif self.next_handler:
            self.next_handler.handle(request)

class Level2Support(SupportHandler):
    def handle(self,request):
        if request=="bug":
            print("Level 2 handled Software Bug")
        elif self.next_handler:
            self.next_handler.handle(request)
class Manager(SupportHandler):
    def handle(self,request):
        if request=="refund":
            print("Manager approved Refund")
        else:
            print("No one could handle the request")
```

## Visitor Pattern

**PS: Different Shapes and different operations on them like area parameters etc**

```python
from abc import ABC, abstractmethod
```

```python
#Visitor
class Visitor(ABC):
    @abstractmethod
    def visit_circle(self,circle):
        pass
    @abstractmethod
    def visit_rectangle(self,rectangle):
        pass
```

```python
#Element
class shape(ABC):
    @abstractmethod
    def accept(self,visitor):
        pass
```

```python
#Concrete Elements
class Circle(shape):
    def __init__(self,radius):
        self.radius=radius
    def accept(self,visitor):
        visitor.visit_circle(self)
class Rectangle(shape):
    def __init__(self,width,height):
        self.width=width
        self.height=height
    def accept(self,visitor):
        visitor.visit_rectangle(self)
```

```python
#Concrete Visitor
class AreaVisitor(Visitor):
    def visit_circle(self,circle):
        area=3.14*circle.radius*circle.radius
        print(f"Circle Area: {area}")
    def visit_rectangle(self,rectangle):
        area=rectangle.width*rectangle.height
        print(f"Rectangle Area: {area}")
```

```python
#Client
shapes=[
    Circle(5),
    Rectangle(4,6)
]
visitor=AreaVisitor()
for shape in shapes:
    shape.accept(visitor)
```

## Interpreter Pattern

**PS: Build a simple interpreter for expressions like 5+3=8**

```python
from abc import ABC, abstractmethod
```

```python
#Abstract Expresssion
class Expression(ABC):
    @abstractmethod
    def interpret(self):
        pass
```

```python
#Terminal Expression
class Number(Expression):
    def __init__(self,value):
        self.value=value
    def interpret(self):
        return self.value
```

```python
#Non-Terminal Expression
class Number(Expression):
    def __init__(self,value):
        self.value=value
    def interpret(self):
        return self.value
```

```python
#Non Terminal Expression
class Add(Expression):
    def __init__(self,left,right):
        self.left=left
        self.right=right
    def interpret(self):
        return self.left.interpret() + self.right.interpret()
```

```python
#Client
five=Number(5)
three=Number(3)
expression=Add(five,three)
print(expression.interpret())
```

```python
expr=Add(
    Add(Number(10),Number(20)),
    Number(5)
)
print(expr.interpret())
```
